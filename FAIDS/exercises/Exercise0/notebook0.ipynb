Let's compute the **variance** of a list manually in python.
For an explanation of the formula see [Wikipedia](https://en.wikipedia.org/wiki/Variance#Definition) or *Lecture 1 - Probability Theory*.

```python
l = [3,1,2,3,2,1,2,1,5,3,2,3] # define a list named "l" with some integers#
```

```python
#TODO: modify the list (or create a new one) using other datatypes
l.pop()
```

```python
#Get mean
mean = sum(l)/len(l) # mean = sum of values divided by number of values (length of list)

#Get squared deviations from mean 
deviations = [] # create empty list
for value in l: # go through every value in the above defined list
    deviation = (mean-value)**2 # compute squared difference between mean and list value
    deviations.append(deviation) # add this deviation to the deviations list

#Get mean of deviations
variance = sum(deviations)/len(deviations) # compute the mean of the deviations 
```

```python
print(variance) # prints the value of the variable "variance" to the notebook output
```
