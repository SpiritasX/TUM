- b) Among 100 000 produced cell phones, 200 have a defect. An inspection procedure detects the correct result in both cases (defect or not defect) with probability of 99%. What is the probability that a phone has indeed a defect, given that the inspection says it has a defect? [4p]

```python
# D - defect
# nD - not defect
# I - positive inspection
# P(D | I) = P(I | D) * P(D) / P(I)
# P(I | D) = 0.99
# P(D) = 200/100000 = 0.002
# P(I) = P(I | D) * P(D) + P(I | nD) * P(nD) = 0.99 * 0.002 + 0.01 * 0.998 = 0.01196

# P(D | I) = 0.99 * 0.002 / 0.01196 = 16.56%
```

## Exercise 1.1 d)
Reference for visualization of the given distribution. Fell free to play around.

```python
%pip install matplotlib
```

```python
import matplotlib.pyplot as plt #for plotting

theta = 0.10 # the distribution parameters
ks = range(0,40) # k values (x-axis)

Ps = [] # probabilities at k (need to be computed, y-axis)
for k in ks:
    P = theta * (1-theta)**(k-1) #apply formula to get P from k and theta
    Ps.append(P) # add P to the list of probabilities

plt.figure() #create new figure (documentation: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.figure.html)
plt.scatter(ks,Ps) #plot points on 2D plane (documentation: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)
plt.xlabel("k")
plt.ylabel("P")
plt.show() #shows figure

```

## Exercise 1.2 **[8p]**

**a)** Modify the given code to generate and plot **100,000 samples** from a normal distribution with **µ = 10** and **σ=2.5**. **[2p]**

```python
import numpy as np

N = 1000

samples = np.random.normal(size=N)

plt.figure()
plt.hist(samples, bins=50)
plt.show()
```

**b)** Compute (e.g. using numpy) and report the mean and standard deviation of the above generated samples. **[1p]**

```python
mean = samples.mean()
standard_deviation = samples.std()

print(f"mean:{mean}")
print(f"standard_deviation:{standard_deviation}")
```

**c)** Recycle the code from a) & b) to sample from an exponential distribution. Use a non-default value for λ. Show your code and the histogram. **[2p]**

```python
N = 1000

l = 0.2

samples = np.random.exponential(scale = 1/l, size=N)

plt.figure()
plt.hist(samples, bins=50)
plt.show()
```
